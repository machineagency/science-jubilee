## Camera Tool Demo

Demonstrates the unified `Camera` tool with a USB webcam (`opencv` backend).
The same class supports `"http"` (remote Pi server) and `"picamera"` (local Pi) backends via config.

Before starting: make sure your USB webcam is connected.

```python
from science_jubilee.Machine import Machine
from science_jubilee.tools.Camera import Camera

%load_ext autoreload
%autoreload 2
```

```python
m = Machine(address='10.18.178.192')
# uncomment the line below to home all axes if you haven't
# m.home_all()
```

### Initialize and Pick Up Camera

The `tool_index` and `tool_name` should match your machine's `config.g`.

```python
# from config file
cam = Camera.from_config(1, 'Camera', 'camera_opencv.json')
# or directly: cam = Camera(1, 'Camera', backend='opencv', camera_index=0)

m.load_tool(cam)
```

```python
# pick up tool when you are ready
m.pickup_tool(cam)
```

### Preview the image
`show_image()` method returns the numpy array, which Jupyter displays as text. The trailing ; suppresses Jupyter's auto-display of the return value. When you need the frame for processing, use frame = cam.show_image() — the assignment also suppresses the raw array output.

```python
frame = cam.show_image()
# this does the same thing:
# cam.show_image();
```

### Save the image
`capture()` saves the image to the default directory defined in the camera config.

```python
cam.capture('test_image.jpg')
```

### Record Video

- `record(filename, duration)` — blocks until the duration is up, then returns
- `start_recording(filename)` — starts recording in the background and returns immediately, so you can move the machine or do other work while recording. Call `stop_recording()` when done. Optionally pass `duration` as a safety limit.

```python
cam.record('test_video.mp4', duration=5)
```

```python
# records in the background until stop_recording() is called
cam.start_recording('sync_video.mp4')
m.move_to(x=100, y=100, z=50)
m.move_to(x=150, y=150, z=100)
cam.stop_recording()
```
